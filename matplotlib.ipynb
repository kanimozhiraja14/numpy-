```python
X = np.array([0,40,20,60,5,70])
X1 = np.array([0,10,30,40,50,20])

plt.plot(X,linewidth="5",marker='o')
plt.plot(X1,linewidth="4",marker='s')
plt.show()
```

```python
X = np.array([0,2,4,6,5,8])
Y = np.array([0,10,15,20,25,30])
plt.plot(X,linewidth="5",marker='o')
plt.plot(X1,linewidth="4",marker='s')
plt.title("X and Y plot")
plt.ylabel("this is Y axis")
plt.xlabel("this is X axis")
plt.show()
```
