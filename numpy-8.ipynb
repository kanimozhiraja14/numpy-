```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5])
np.save("data.npy",arr)
```

```python
read = np.load("data.npy")
print(read)
```

```python
brr = np.array([6,7,8,9])
np.savez("data.npz",arr,brr)
```

```python
read = np.load ("data.npz")
```

```python
print(read)
```

```python
print(read["arr_0"])
```

```python
print(read["arr_1"])
```
