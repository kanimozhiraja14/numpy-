```python
import numpy as np
import pandas as pd
import seaborn as  sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_openml
```

```python
titanic_data = fetch_openml("titanic", version = 1,as_frame = True)
```

```python
df = titanic_data["data"]
```

```python
df['survived'] = titanic_data['target']
```

```python
df.head()
```

```python
df.drop(["name","ticket","home.dest"],axis =1,inplace = True)
df.head()
```

```python
df.drop(["cabin","body","boat"],axis =1,inplace=True)
df.head()
```

```python
sex = pd.get_dummies(df['sex'], drop_first = True)
print(sex)
```

```python
df['sex'] = sex
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imp_mean = SimpleImputer(strategy ='mean')
```

```python
df['age'] = imp_mean.fit_transform(df[['age']])
df['fare'] = imp_mean.fit_transform(df[['fare']])
```

```python
df.isnull().sum()
```

```python
imp_freq = SimpleImputer(strategy = 'most_frequent')
```

```python
df['embarked']=imp_freq.fit_transform(df[['embarked']]).ravel()
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
embar= pd.get_dummies(df['embarked'],drop_first = True)
print(embar)
```

```python
df.drop(['embarked'],axis =1,inplace=True)
df= pd.concat([df,embar],axis =1) 
```

```python
df.head()
```

```python
X =df.drop(['survived'],axis =1)
df.head()
```

```python
y =df['survived']
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size =0.3, random_state=1)
```

```python
print(X_train.shape,y_train.shape)
print(X_test.shape,y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
mod =LogisticRegression(max_iter=1000)
mod.fit(X_train,y_train)
```

```python
pred = mod.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
from sklearn.metrics  import confusion_matrix
confusion_matrix(y_test,pred)
```
